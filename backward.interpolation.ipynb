```python
import numpy as np

# Given data
x = np.array([1, 2, 3, 4, 5])
y = np.array([1, 8, 27, 64, 125])

n = len(x)
h = x[1] - x[0]

# Difference table
D = np.zeros((n, n))
D[:, 0] = y

for j in range(1, n):
    for i in range(n-j):
        D[i, j] = D[i+1, j-1] - D[i, j-1]

# Interpolation point
xp = 4.5
u = (xp - x[n-1])/h

sum = D[n-1, 0]
p = 1

for i in range(1, n):
    p = p * (u+i-1)/i
    sum = sum + p*D[n-i-1, i]

print("Interpolated value =", sum)
```
