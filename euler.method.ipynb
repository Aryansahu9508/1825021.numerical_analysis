```python
import numpy as np
import matplotlib.pyplot as plt

# Initial values
x0 = 0
y0 = 1
h = 0.1
xn = 2

def f(x, y):
    return x + y

# Arrays
x = np.arange(x0, xn+h, h)
y = np.zeros(len(x))

y[0] = y0

# Euler Method
for i in range(len(x)-1):
    y[i+1] = y[i] + h*f(x[i], y[i])

# Print results
for i in range(len(x)):
    print(x[i], y[i])

# Plot graph
plt.plot(x, y, 'o-', label='Euler Method')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Euler Method')
plt.legend()
plt.grid()
plt.show()
```
