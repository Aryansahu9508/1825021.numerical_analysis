```python
import numpy as np

# Differential equations
def f(x, y, z):
    return z

def g(x, y, z):
    return -y

# RK4 Method
def rk4(s):
    x = 0
    y = 0
    z = s
    xn = np.pi/2
    h = 0.01

    while x < xn:
        step = min(h, xn-x)

        k1 = step*f(x, y, z)
        l1 = step*g(x, y, z)

        k2 = step*f(x+step/2, y+k1/2, z+l1/2)
        l2 = step*g(x+step/2, y+k1/2, z+l1/2)

        k3 = step*f(x+step/2, y+k2/2, z+l2/2)
        l3 = step*g(x+step/2, y+k2/2, z+l2/2)

        k4 = step*f(x+step, y+k3, z+l3)
        l4 = step*g(x+step, y+k3, z+l3)

        y = y+(k1+2*k2+2*k3+k4)/6
        z = z+(l1+2*l2+2*l3+l4)/6

        x = x+step

    return y

# Shooting Method
s1 = 0.5
s2 = 1.5

for i in range(10):
    y1 = rk4(s1)
    y2 = rk4(s2)

    s3 = s2 + (1-y2)*(s2-s1)/(y2-y1)

    if abs(s3-s2) < 1e-10:
        s2 = s3
        break

    s1 = s2
    s2 = s3

# Final result
print("Initial slope =", s2)
print("Final value =", rk4(s2))
```
