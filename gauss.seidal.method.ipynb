```python
import numpy as np

# Coefficient matrix
A = np.array([[10, 1, 1],
              [2, 10, 1],
              [2, 2, 10]])

# Constant vector
B = np.array([12, 13, 14])

# Initial guess
x = np.zeros(3)

# Number of iterations
n = 10

# Gauss Seidel Method
for k in range(n):
    for i in range(3):
        s = 0

        for j in range(3):
            if i != j:
                s = s + A[i,j]*x[j]

        x[i] = (B[i]-s)/A[i,i]

# Print solution
print("Solution:")
print(x)
```
