```python
p, q, r = 1, 1, 1
x0, xf = 0, 2
y0 = -1
yf_target = 5

n = 100
h = (xf - x0) / n


t0 = 0.0
t1 = 1.0

# --- Solve for Guess 1 (t0) ---
x, y1, y2 = x0, y0, t0
for _ in range(n):
    k1_y1 = y2
    k1_y2 = p - q*y2 - r*y1**2
    k2_y1 = y2 + (h/2)*k1_y2
    k2_y2 = p - q*(y2 + (h/2)*k1_y2) - r*(y1 + (h/2)*k1_y1)
    k3_y1 = y2 + (h/2)*k2_y2
    k3_y2 = p - q*(y2 + (h/2)*k2_y2) - r*(y1 + (h/2)*k2_y1)
    k4_y1 = y2 + h*k3_y2
    k4_y2 = p - q*(y2 + h*k3_y2) - r*(y1 + h*k3_y1)

    y1 += (h/6)*(k1_y1 + 2*k2_y1 + 2*k3_y1 + k4_y1)
    y2 += (h/6)*(k1_y2 + 2*k2_y2 + 2*k3_y2 + k4_y2)
    x += h
f0 = y1 - yf_target


x, y1, y2 = x0, y0, t1
for _ in range(n):
    k1_y1 = y2
    k1_y2 = p - q*y2 - r*y1
    k2_y1 = y2 + (h/2)*k1_y2
    k2_y2 = p - q*(y2 + (h/2)*k1_y2) - r*(y1 + (h/2)*k1_y1)
    k3_y1 = y2 + (h/2)*k2_y2
    k3_y2 = p - q*(y2 + (h/2)*k2_y2) - r*(y1 + (h/2)*k2_y1)
    k4_y1 = y2 + h*k3_y2
    k4_y2 = p - q*(y2 + h*k3_y2) - r*(y1 + h*k3_y1)

    y1 += (h/6)*(k1_y1 + 2*k2_y1 + 2*k3_y1 + k4_y1)
    y2 += (h/6)*(k1_y2 + 2*k2_y2 + 2*k3_y2 + k4_y2)
    x += h
f1 = y1 - yf_target  # Error at xf


correct_slope = t1 - f1 * (t1 - t0) / (f1 - f0)
print(f"The exact required initial slope y'(0) is: {correct_slope}")


x, y1, y2 = x0, y0, correct_slope
x_result = [x]
y_result = [y1]

for _ in range(n):
    k1_y1 = y2
    k1_y2 = p - q*y2 - r*y1
    k2_y1 = y2 + (h/2)*k1_y2
    k2_y2 = p - q*(y2 + (h/2)*k1_y2) - r*(y1 + (h/2)*k1_y1)
    k3_y1 = y2 + (h/2)*k2_y2
    k3_y2 = p - q*(y2 + (h/2)*k2_y2) - r*(y1 + (h/2)*k2_y1)
    k4_y1 = y2 + h*k3_y2
    k4_y2 = p - q*(y2 + h*k3_y2) - r*(y1 + h*k3_y1)

    y1 += (h/6)*(k1_y1 + 2*k2_y1 + 2*k3_y1 + k4_y1)
    y2 += (h/6)*(k1_y2 + 2*k2_y2 + 2*k3_y2 + k4_y2)
    x += h

    x_result.append(x)
    y_result.append(y1)

print(f"Final targeting check: y({x}) = {y1} (Target was {yf_target})")
```
