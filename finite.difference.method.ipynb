```python
import numpy as np
import matplotlib.pyplot as plt

# Define values
n = 10
a = 0
b = 1
h = (b-a)/n

# Create x array
x = np.linspace(a, b, n+1)

# Create matrix A
A = np.zeros((n-1, n-1))

for i in range(n-1):
    A[i,i] = -2

    if i > 0:
        A[i,i-1] = 1

    if i < n-2:
        A[i,i+1] = 1

B = np.ones(n-1)*(-2*h**2)

Y = np.linalg.solve(A, B)

# Apply boundary conditions
y = np.zeros(n+1)
y[1:n] = Y

# Print values
print("X values:", x)
print("Y values:", y)

# Plot graph
plt.plot(x, y, 'o-', label="Finite Difference Method")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Finite Difference Method")
plt.legend()
plt.grid()
plt.show()
```
