```python
import numpy as np

# Given matrix
A = np.array([[4, 1, 0],
              [1, 3, 1],
              [0, 1, 2]])

x = np.array([1, 1, 1], dtype=float)

# Number of iterations
n = 10
for i in range(n):

    y = np.dot(A, x)

    eigenvalue = np.max(np.abs(y))

    x = y / eigenvalue

print("Dominant Eigenvalue =", eigenvalue)
print("Eigenvector =", x)
```
