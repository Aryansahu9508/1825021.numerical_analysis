```python
# -runge kutta method -*-

import numpy as np
import matplotlib.pyplot as plt

def f(x,y):
     return x**3 - y**2 + 3

x0 = float(input("enter initial value of x:"))
y0 = float(input("enter initial value of y:"))
h = float(input("enter the step size:"))
xn = float(input("enter the final value of x:"))
x_values = [x0]
y_value = [y0]

x = x0
y = y0

while x<=xn:
    k1 = h*f(x,y)
    k2 = h*f(x + h/2 , y+ k1/2)
    k3 = h*f(x+h/2 ,y + k2/2)
    k4 = h*f(x+h,y+k3)

    y = y + (k1 + 2*k2 + 2*k3 + k4)/6
    x = x+h

    x_values.append(x)
    y_value.append(y)

    for i in range(len(x_values)):
        print(f)

plt.plot(x_values,y_value)
plt.xlabel("x")
plt.ylabel("y")
plt.title("runge kutta method")
plt.grid(True)
plt.show()
```
